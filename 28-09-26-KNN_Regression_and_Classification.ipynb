```python
# ============================================================
# KNN REGRESSION - COMPLETE CODE
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ------------------------------------------------------------
# 1. Create Dummy Dataset
# ------------------------------------------------------------

df = pd.DataFrame()

df['age'] = np.random.randint(18, 70, 100)

df['salary'] = np.random.randint(
    100000,
    200000,
    100
)

df['#ofmembers'] = np.random.randint(
    1,
    8,
    100
)

# EMI is continuous because this is regression
df['emi'] = np.random.randint(
    5000,
    50000,
    100
)


print("Dataset:")
print(df.head())


# ------------------------------------------------------------
# 2. Independent Variables
# ------------------------------------------------------------

x = df[
    [
        'age',
        'salary',
        '#ofmembers'
    ]
]


# ------------------------------------------------------------
# 3. Dependent Variable
# ------------------------------------------------------------

y = df['emi']


# ------------------------------------------------------------
# 4. Train Test Split
# ------------------------------------------------------------

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)


# ------------------------------------------------------------
# 5. KNN Regression Function
# ------------------------------------------------------------

def neighbors_algo(n):

    # Create KNN Regressor
    kn = KNeighborsRegressor(
        n_neighbors=n
    )


    # Train the model
    kn.fit(
        x_train,
        y_train
    )


    # Prediction
    y_pred = kn.predict(x_test)


    # --------------------------------------------------------
    # Regression Metrics
    # --------------------------------------------------------

    # Mean Absolute Error
    mae = mean_absolute_error(
        y_test,
        y_pred
    )


    # Mean Squared Error
    mse = mean_squared_error(
        y_test,
        y_pred
    )


    # Root Mean Squared Error
    rmse = np.sqrt(mse)


    # R2 Score
    r2 = r2_score(
        y_test,
        y_pred
    )


    # --------------------------------------------------------
    # Print Results
    # --------------------------------------------------------

    print("\nNeighbors :", n)

    print("MAE  :", mae)

    print("MSE  :", mse)

    print("RMSE :", rmse)

    print("R2   :", r2)


    # Return Results
    return (
        "Neighbors " + str(n),
        mae,
        mse,
        rmse,
        r2
    )


# ------------------------------------------------------------
# 6. Test Different K Values
# ------------------------------------------------------------

neighbors_algo(3)

neighbors_algo(5)

neighbors_algo(7)

neighbors_algo(9)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix


# ============================================================
# 1. CREATE DUMMY DATASET
# ============================================================

np.random.seed(42)

df = pd.DataFrame()

df['age'] = np.random.randint(18, 70, 100)

df['salary'] = np.random.randint(100000, 200000, 100)

df['#ofmembers'] = np.random.randint(1, 8, 100)

# Target
# 0 = No
# 1 = Yes
df['emi'] = np.random.randint(0, 2, 100)


print("Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nNull Values:")
print(df.isnull().sum())


# ============================================================
# 2. DEFINE X AND Y
# ============================================================

x = df[
    [
        'age',
        'salary',
        '#ofmembers'
    ]
]

y = df['emi']


# ============================================================
# 3. TRAIN TEST SPLIT
# ============================================================

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 4. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)

x_test = scaler.transform(x_test)


# ============================================================
# 5. CREATE KNN CLASSIFIER
# ============================================================

kn = KNeighborsClassifier()


# ============================================================
# 6. GRID SEARCH PARAMETERS
# ============================================================

parameters = {

    'n_neighbors': [
        3,
        5,
        7,
        9,
        11
    ],

    'weights': [
        'uniform',
        'distance'
    ],

    'metric': [
        'euclidean',
        'manhattan',
        'minkowski'
    ],

    'p': [
        1,
        2
    ]
}


# ============================================================
# 7. CREATE GRIDSEARCHCV
# ============================================================

grid = GridSearchCV(
    estimator=kn,
    param_grid=parameters,
    cv=5,
    scoring='accuracy'
)


# ============================================================
# 8. TRAIN GRIDSEARCH
# ============================================================

grid.fit(
    x_train,
    y_train
)


# ============================================================
# 9. BEST PARAMETERS
# ============================================================

print("\nBest Parameters:")

print(
    grid.best_params_
)


# ============================================================
# 10. BEST CROSS VALIDATION SCORE
# ============================================================

print("\nBest Cross Validation Accuracy:")

print(
    grid.best_score_
)


# ============================================================
# 11. GET BEST MODEL
# ============================================================

best_kn = grid.best_estimator_

print("\nBest KNN Model:")

print(best_kn)


# ============================================================
# 12. PREDICTION
# ============================================================

y_pred = best_kn.predict(
    x_test
)


print("\nActual Values:")

print(
    y_test.values
)


print("\nPredicted Values:")

print(
    y_pred
)


# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test.values,
    y_pred
)

print("\nConfusion Matrix:")

print(cm)


# ============================================================
# 14. EXTRACT TN, FP, FN, TP
# ============================================================

TN, FP, FN, TP = cm.ravel()


print("\nTN =", TN)

print("FP =", FP)

print("FN =", FN)

print("TP =", TP)


# ============================================================
# 15. CALCULATE METRICS MANUALLY
# ============================================================

# Accuracy
accuracy = (
    (TP + TN)
    /
    (TP + TN + FP + FN)
)


# Precision
precision = (
    TP
    /
    (TP + FP)
)


# Recall / Sensitivity
recall = (
    TP
    /
    (TP + FN)
)


# Specificity
specificity = (
    TN
    /
    (TN + FP)
)


# F1 Score
f1 = (
    2 * precision * recall
    /
    (precision + recall)
)


# False Positive Rate
fpr = (
    FP
    /
    (FP + TN)
)


# False Negative Rate
fnr = (
    FN
    /
    (FN + TP)
)


# Negative Predictive Value
npv = (
    TN
    /
    (TN + FN)
)


# ============================================================
# 16. PRINT ALL METRICS
# ============================================================

print("\n========== FINAL MODEL METRICS ==========")

print("Accuracy    :", accuracy)

print("Precision   :", precision)

print("Recall      :", recall)

print("Specificity :", specificity)

print("F1 Score    :", f1)

print("FPR         :", fpr)

print("FNR         :", fnr)

print("NPV         :", npv)


# ============================================================
# 17. PRINT METRICS AS PERCENTAGE
# ============================================================

print("\n========== METRICS IN PERCENTAGE ==========")

print(
    "Accuracy    :",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Precision   :",
    round(precision * 100, 2),
    "%"
)

print(
    "Recall      :",
    round(recall * 100, 2),
    "%"
)

print(
    "Specificity :",
    round(specificity * 100, 2),
    "%"
)

print(
    "F1 Score    :",
    round(f1 * 100, 2),
    "%"
)


# ============================================================
# 18. PREDICT NEW CUSTOMER
# ============================================================

new_customer = pd.DataFrame(
    {
        'age': [35],
        'salary': [150000],
        '#ofmembers': [4]
    }
)


# Apply same scaler
new_customer_scaled = scaler.transform(
    new_customer
)


# Make prediction
prediction = best_kn.predict(
    new_customer_scaled
)


print("\nNew Customer Prediction:")

print(
    prediction[0]
)


if prediction[0] == 1:

    print("Customer belongs to EMI = YES")

else:

    print("Customer belongs to EMI = NO")
```
